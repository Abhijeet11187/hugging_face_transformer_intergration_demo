```python
# Contextual Embedding and semantic search engines with transformer model
```

```python
documents = [
  "Quantum mechanics describes the behavior of very small particles.",
  "Photosynthesis is the process by which green plants make food using sunlight.",
  "Shakespeare's plays are a testament to English literature.",
  "Artificial Intelligence aims to create machines that can think and learn.",
  "The pyramids of Egypt are historical monuments that have stood for thousands of years.",
  "Biology is the study of living organisms and their interactions with the environment.",
  "Music therapy can aid in the mental well-being of individuals.",
  "The Milky Way is just one of billions of galaxies in the universe.",
  "Economic theories help understand the distribution of resources in society.",
  "Yoga is an ancient practice that involves physical postures and meditation."
]
```

```python
#!pip install -U sentence-transformers
```

```python
# !nvidia-smi
```

```python
from sentence_transformers import SentenceTransformer, util
import torch
```

```python
model=SentenceTransformer('all-MiniLM-L6-v2')
```

```python
document_embedding=model.encode(documents)
```

```python
new_text="What is Generative AI?"
```

```python
new_text_embedding=model.encode(new_text)
```

```python
# Checking cosine similarity
```

```python
cos_scores=util.pytorch_cos_sim(new_text_embedding,document_embedding)
```

```python
cos_scores
```

```python
# Here fourth cosine similarity is closer 0.59
```

```python
top_result=torch.topk(cos_scores,k=1)
```

```python
top_result
```

```python
idx=top_result.indices[0]
idx
```

```python
documents[idx]
```

```python
# Create a function to return the top similar documents based on any query
```

```python
def semantic_search_engine(query,embedded_model):
  query_embedding=embedded_model.encode(query)
  cos_scores=util.pytorch_cos_sim(query_embedding,document_embedding)[0]
  top_result=torch.topk(cos_scores,k=1)
  idx=top_result.indices[0]
  return documents[idx]
```

```python
sentence_one="Tell me about AI"
```

```python
semantic_search_engine(sentence_one,model)
```

```python
sentence_two="Do you know what are Pyramids"
```

```python
semantic_search_engine(sentence_two,model)
```
